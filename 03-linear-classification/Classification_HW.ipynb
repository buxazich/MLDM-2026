#### About the data
Here's some of the columns
* Name - a string with person's full name
* Survived - 1 if a person survived the shipwreck, 0 otherwise.
* Pclass - passenger class. Pclass == 3 is cheap'n'cheerful, Pclass == 1 is for moneybags.
* Sex - a person's gender
* Age - age in years, if available
* SibSp - number of siblings on a ship
* Parch - number of parents on a ship
* Fare - ticket cost
* Embarked - port where the passenger embarked
 * C = Cherbourg; Q = Queenstown; S = Southampton

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, RobustScaler, PolynomialFeatures, OneHotEncoder
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.impute import SimpleImputer

def feature_selection_and_preprocessing(dataset):
  # E.g.:
  features = dataset[["Fare", "Age", "Sex", "Pclass"]].copy()
  features['Embarked'] = dataset.Embarked.fillna('unknown')

  # Reduced the effect of very expensive tickets.
  features['Fare'] = np.log1p(features['Fare'])

  # Extracted titles and grouped uncommon ones together.
  titles = dataset.Name.str.extract(r',\s*([^.]*)\.', expand=False)
  titles = titles.replace({'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs'})
  features['Title'] = titles.where(titles.isin(['Mr', 'Mrs', 'Miss', 'Master']), 'Rare')

  # Grouped passengers by family size.
  family_size = dataset.SibSp + dataset.Parch + 1
  features['FamilyGroup'] = pd.cut(family_size, [0, 1, 4, 20], labels=['alone', 'small', 'large']).astype(str)

  # Marked missing ages and known cabins, and combined sex with class.
  features['HasCabin'] = dataset.Cabin.notna().astype(int)
  features['AgeMissing'] = dataset.Age.isna().astype(int)
  features['SexClass'] = dataset.Sex + '_' + dataset.Pclass.astype(str)

  # Added surnames and tickets to identify related passengers.
  features['Surname'] = dataset.Name.str.split(',').str[0]
  features['Ticket'] = dataset.Ticket

  # Combined the family size group with sex.
  features['FamilySex'] = features.FamilyGroup + '_' + dataset.Sex
  return features

model = make_pipeline(
    # E.g.
    make_column_transformer(
        (OneHotEncoder(sparse_output=False, handle_unknown='ignore'),
         ['Sex', 'Embarked', 'Title', 'FamilyGroup', 'SexClass', 'Surname', 'Ticket', 'Pclass', 'FamilySex']),
        # Learned medians and scaling values from the training set only.
        (make_pipeline(SimpleImputer(strategy='median'), StandardScaler()),
         ['Fare', 'Age', 'HasCabin', 'AgeMissing']),
        remainder='passthrough'),
    # Allowed more iterations so the model could converge.
    # Used C=3 to reduce regularization.
    LogisticRegression(C=3, max_iter=1000))


# Validation code (do not touch)
data = pd.read_csv("train.csv", index_col='PassengerId')
data_train, data_test = train_test_split(data, test_size=200, random_state=42)

model.fit(
    feature_selection_and_preprocessing(
        data_train.drop('Survived', axis=1)
    ),
    data_train['Survived']
)

train_predictions = model.predict(
    feature_selection_and_preprocessing(
        data_train.drop('Survived', axis=1)
    )
)

test_predictions = model.predict(
    feature_selection_and_preprocessing(
        data_test.drop('Survived', axis=1)
    )
)

print("Train accuracy:", accuracy_score(
    data_train['Survived'],
    train_predictions
))
print("Test accuracy:", accuracy_score(
    data_test['Survived'],
    test_predictions
))
```
